```python
agenda = {}

def adicionar_contato(nome,telefone):
  agenda[nome] = telefone

def lista_de_contatos():
  for nome_contatos in agenda:
    print(nome_contatos, ':' ,agenda[nome_contatos])

def busca_contato(nome):
  if nome in agenda:
    print(f'nome: {nome} telefone: {agenda[nome]}')
  else:
    print('contato nao encontrado')

def remover_contato(nome):
 if nome in agenda:
  del agenda[nome]


adicionar_contato('fernando',9939393939)
adicionar_contato('gomes',9939393939)

lista_de_contatos()
busca_contato('fernando')
remover_contato('fernando')
lista_de_contatos()
```
